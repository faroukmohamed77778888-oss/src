add item


```python
class Task1:

    def __init__(self, my_tuple, item):
        self.my_tuple = my_tuple
        self.item = item

    def add_item(self):
        x = list(self.my_tuple)
        x.append(self.item)
        return tuple(x)


my_tuple = (1, 2, 3)

obj = Task1(my_tuple, 4)

print(obj.add_item())
```

sum all item


```python
class Task2:

    def __init__(self, numbers):
        self.numbers = numbers

    def sum_numbers(self):
        total = 0

        for number in self.numbers:
            total += number

        return total


numbers = [5, 10, 15, 20]

obj = Task2(numbers)

print(obj.sum_numbers())
```

```python
class Task4:

    def __init__(self, numbers):
        self.numbers = numbers

    def smallest(self):
        small = self.numbers[0]

        for number in self.numbers:
            if number < small:
                small = number

        return small


numbers = [10, 3, 25, 1, 7]

obj = Task4(numbers)

print(obj.smallest())
```

```python
class Task5:

    def __init__(self, numbers):
        self.numbers = numbers

    def largest(self):
        large = self.numbers[0]

        for number in self.numbers:
            if number > large:
                large = number

        return large


numbers = [10, 3, 25, 1, 7]

obj = Task5(numbers)

print(obj.largest())
```

```python
class Task8:

    def __init__(self, my_set, item):
        self.my_set = my_set
        self.item = item

    def remove(self):
        self.my_set.remove(self.item)
        return self.my_set


my_set = {1, 2, 3, 4, 5}

obj = Task8(my_set, 3)

print(obj.remove())
```

```python
class Task9:

    def __init__(self, set1, set2):
        self.set1 = set1
        self.set2 = set2

    def check(self):
        return self.set1.issubset(self.set2)


set1 = {1, 2}
set2 = {1, 2, 3, 4}

obj = Task9(set1, set2)

print(obj.check())
```

```python
class Task3:

    def __init__(self, numbers):
        self.numbers = numbers

    def multiply(self):
        result = 1

        for number in self.numbers:
            result *= number

        return result


numbers = [2, 3, 4]

obj = Task3(numbers)

print(obj.multiply())
```

```python
class Task6:

    def __init__(self, words):
        self.words = words

    def count(self):
        count = 0

        for word in self.words:
            if len(word) >= 2:
                count += 1

        return count


words = ["hi", "a", "hello", "no", "on"]

obj = Task6(words)

print(obj.count())
```
