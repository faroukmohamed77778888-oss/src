```python
my_set = {10, 3, 25, 1, 7}

numbers = list(my_set)

maximum = numbers[0]
minimum = numbers[0]

for number in numbers:
    if number > maximum:
        maximum = number

    if number < minimum:
        minimum = number

print("Max:", maximum)
print("Min:", minimum)
```
